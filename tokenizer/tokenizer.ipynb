```python
# Run this in a notebook to see all unique SELFIES tokens in ZINC250k
import selfies as sf
import pandas as pd
from collections import Counter

df = pd.read_csv("hf://datasets/edmanft/zinc250k/zinc250k_selfies.csv")

all_tokens = []
for selfies_str in df["selfies"]:
    tokens = list(sf.split_selfies(selfies_str))
    all_tokens.extend(tokens)

token_counts = Counter(all_tokens)
print(f"Unique SELFIES tokens: {len(token_counts)}")
print("\nAll tokens:")
for token, count in sorted(token_counts.items(), key=lambda x: -x[1]):
    print(f"  {token}: {count:,}")
```

```python
# Run this to understand your sequence lengths
import selfies as sf
import pandas as pd

df = pd.read_csv("hf://datasets/edmanft/zinc250k/zinc250k_selfies.csv")

lengths = [len(list(sf.split_selfies(s))) for s in df["selfies"]]

print(f"Min length  : {min(lengths)}")
print(f"Max length  : {max(lengths)}")
print(f"Mean length : {sum(lengths)/len(lengths):.1f}")
print(f"95th pctile : {sorted(lengths)[int(0.95*len(lengths))]}")
print(f"99th pctile : {sorted(lengths)[int(0.99*len(lengths))]}")
```
